### ***Python Matplotlib***  

Matplotlib is a comprehensive, cross-platform Python library for creating static, animated, interactive, and publication-quality data visualizations

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.figure()
plt.plot()
plt.show();
```

***Scatter Plot***

```python
x = np.array([1,2,3])
y = np.array([1,2,3])

plt.scatter(x,y,c='r',marker ="^", s=100)
plt.show();
```

```python
x = np.array([1,2,3,4])
y = np.array([1,2,3,4])
color = ["blue","yellow","red","green"]
plt.scatter(x,y,c=color,marker ="^", s=100)
plt.show();
```

***Line graph***

```python
x = np.array([1,2,3,5,4])
y = np.array([1,2,3,5,3])
color = ["red","yellow","green","blue","black"]
plt.plot(x,y,c='green', linestyle="dashed")
plt.scatter(x,y,c=color,marker ="^", s=100)
plt.show();
```

***Bar chart***

```python
year =["1900","1905","1910","1915","1920","1925","1930"]
lang = ["None", "English","Hindi","Tamil","Japanes","siwiz","Egpyt"]
c = ["Green", "orange","red","yellow","blue","black","violet"]

plt.bar(year,lang,color=c,align="edge")
plt.show();
```

```python
#frequency distribution

np.random.seed(2)
a = np.random.normal(80,5,100)
plt.hist(a,color="pink",bins=10,edgecolor="black")
plt.show();
```

***Pie Chart***

```python
year =[25,20,5,10,15,12.5,12.5]
exp =[0.1,0.2,0.1,0.1,0.1,0.1,0.2]
lang = ["None", "English","Hindi","Tamil","Japanes","siwiz","Egpyt"]
c = ["Green", "orange","red","yellow","blue","black","violet"]

plt.pie (year,labels = lang,explode = exp, colors=c)
plt.show();
```

***Title***

```python
year =["1900","1905","1910","1915","1920","1925","1930"]
lang = ["None", "English","Hindi","Tamil","Japanes","siwiz","Egpyt"]
c = ["Green", "orange","red","yellow","blue","black","violet"]

plt.title("Language Known")
plt.xlabel("Years")
plt.ylabel("Languages")
plt.bar(year,lang,color=c,align="edge")
plt.show();
```

```python
#Stack market
stockA = [1,2,3,4,5]
stockB = [6,7,6,9,1]
stockC = [6,1,8,4,9]

plt.title("Stock Market")
plt.plot(stockA, label="stock1",marker="*")
plt.plot(stockB, label="stock2",marker="p")
plt.plot(stockC, label="stock3",marker="o")
plt.legend()
plt.show();
```

***1st method (pyplot api)***

```python
fig = plt.figure()
ax = fig.add_subplot()

ax.plot([1,2,3,5],[-1,2,3,4], c="g")
plt.show();
```

***2nd method (object oriented api)***

```python
fig,ax = plt.subplots()
ax.plot([1,2,3,4],[1,2,4,5])
plt.show();
```

***Linspace***

```python
x = np.linspace(0,10,100)
fig, ax = plt.subplots()
ax.scatter(x,np.exp(x), c='r')
```

***subplots comparing (age vs chol) , (age vs thalach)***

```python
fig, ax = plt.subplots(nrows=2,ncols=1,figsize=(12,8))

scatter = ax[0].scatter(df['age'],df["chol"],c=df['target'],cmap="winter")

ax[0].set(title ="Age vs Cholestrol",xlabel="Age", ylabel="Cholestrol")
ax[0].legend(*scatter.legend_elements(),title="Target")
ax[0].axhline(df['chol'].mean(),color="r",linestyle="dashed")

scatter = ax[1].scatter(df['age'],df["thalach"],c=df['target'],cmap="viridis")

ax[1].set(title ="Age vs Thalach",xlabel="Age", ylabel="thalach")
ax[1].legend(*scatter.legend_elements(),title="Target")
ax[1].axhline(df['thalach'].mean(),color="r",linestyle="dashed")

plt.subplots_adjust(hspace=0.5)
plt.show();
```

***Create a Plot***

```python
fig, ax = plt.subplots(figsize=(10,5))

scatter = ax.scatter(df['age'],df["chol"],c=df['target'],cmap="winter")

ax.set(title ="Heart disease and Cholestrol levels",xlabel="Age", ylabel="Cholestrol")
ax.legend(*scatter.legend_elements(),title="Target")
ax.axhline(df['chol'].mean(),color="r",linestyle="dashed")
plt.savefig("Probability_of_heart_attack.png")
```
