### ***Classe & Objects***  

**Class:** - A blueprint or template for creating a objects. it defines the attributes and behaviors that the objects will have.  

**Object:** - A concrete instance of a class.while the class defines the structure, the object contains actual data.

```python
class techno:
    m = 100 #class variables
    f = 200
obj0 = techno()
print(obj0.m, obj0.f)

obj0.m="Mohamed" #change of variables

print(obj0.m, obj0.f)
```

```python
class techno:
    m = 10
    f = 20
obj1 = techno()
obj2 = techno()

obj1.m = "Mohamed"
print(obj1.m, obj1.f)
print(obj2.m, obj2.f)
```

```python
class techno:
    m = 100
    f = 200

    def mytech(self): #self allows individual objects to access, store, and modify thier own unique data and methods 
        print(self)
obj0 = techno()
print(obj0)
```

```python
class techno:
    m = 10
    f = 20

    def mytech(self, x, y):
        print(f"The function is = {x}, {y}")

obj0 = techno()
print(obj0.m, obj0.f)
obj0.mytech(50, 40)
```

### ***Constructor Function***   

*Syntax*  
**\_\_init\_\_()**

```python
class techno():
    def __init__(self,m,f):
        print("Constuctor Call")
        self.m = m
        self.f = f
obj0 = techno(100, 200)
print(obj0.m, obj0.f)
```

```python
class Dog:
    # The initializer (constructor) method
    def __init__(self, name, breed):
        self.name = name    # self.name creates a unique variable for this instance
        self.breed = breed  # self.breed saves the breed to this instance

    # An instance method
    def bark(self):
        # self is used to retrieve the specific name of the dog calling the method
        print(f"{self.name} says Woof!")

# 1. Create two distinct instances (objects)
dog1 = Dog("Buddy", "Golden Retriever")
dog2 = Dog("Luna", "Husky")

# 2. Call the methods
dog1.bark()  # Output: Buddy says Woof!
dog2.bark()  # Output: Luna says Woof!
```

### ***Destuctors Function***  

*Syntax*  
**\_\_del\_\_()**

```python
class techno:
    def __init__(self,name,mark):
        print("Constuctor Call")
        self.m = name
        self.f = mark
    def __del__(self):
        print(f"Destructors Excuted on {self.m}")
obj0 = techno("Faizal", 200)
print(obj0.m, obj0.f)
obj1 = techno("Aravind", 100)
print(obj1.m, obj1.f)
obj2 = techno("Harish", 120)
print(obj2.m, obj2.f)
print("==========================")
del obj0 
print("==========================")
print(obj0.m, obj0.f)
```
