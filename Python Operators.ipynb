****Operators****
* Arithmetic operators:  +, -, *, / ,** ,//,%
* Assignment operators: =, +=, -=, /=, *= , **=, //=, %=
* Comparison operators: ==, !=, >, <, <=, >=
* Logical operators: and, or, not 
* Identity operators: is, is not
* Membership operators: in, not in


****Arithmetic operators****

```python
a = 50
b = 23

print (a + b)  #addition
print (a - b)  #subtraction
print (a * b)  #Multiplication
print (a / b)  #divison
print (a % b)  #precentage
print (a ** b) #Power
print (a // b) #Quotient
print (a % b)  #Remainder
```

****Assignment Operators****

```python
a = 10 
print (a)

a += 5 
print (a)

a *= 5
print (a)
```

****Comparison Operators****

```python
a = 50
b = 30

print(a == b) 
print (a != b)
print (a < b)
print (a <= b)
print (a > b)
print (a >= b)
```

****Logical Operators****

```python
a = 50
b = 50

print ('-----AND-----')
print(a == b and a == b)
print(a > b and a >= b)
print (a <= b and a < b)        
print (a != b and a!=b ) 

print ('-----OR-----')
print(a == b or a == b) 
print(a > b or a >= b)
print (a <= b or a < b)
print (a != b or a!=b )

print ('-----NOT-----')
print(not(a == b))  
print(not(a > b))
```

****Identity Operators****  


```python
a = 50
b = 50
print ("------is not-----")
print (a != b) # "!=" => value
print (a is not b) # "is not" => object identity

a1 = (1, 2, 3)
b1 = (1, 2, 3)
print ("------is-----")
print (a1 != b1)
print (a1 is b1)
```

****Membership Operators****

```python
a = ["apple", "banana", "cherry", "kiwi"]
print ("------in-----")
print ("apple" in a)
print ("------not in-----")
print ("banana" not in a)
```

****Python Collection****
* *List* - List is a collection of variables which is ordered and changeable. Allows duplicate members. Denoted by **[]**

```python
mylist = ["apple", "banana", "cherry", "kiwi", "mango", "10", "20", "30", "40", "50" ]
print (mylist, type(mylist))
print (mylist[4])
mylist[6] = "15"
print (mylist)
```

* Tuple - Tuple is also a collection of variables, which is ordered and unchangeable. Allows duplicate members.  Denoted by **()**

```python
mytuple = ("apple", "banana", "cherry", "kiwi", "mango", "10", "20", "30", "40", "50" )
print (mytuple, type(mytuple))
print (mytuple[4])
print (mytuple[6])

mytuple[6] = "15"
print (mytuple)
```

****Dictionary****
* Dictionary is a collection that is unordered, changeable and stores data in a key-value pairs. Keys must be unqiue without duplicate keys. It is denoted by curly braces **{}** using a format ***{key: value}***

```python
mydict = {"brand": "Ford", "model": "Mustang", "year": 1964, "price": 20000}

print (mydict, type(mydict))

print (mydict["brand"])

mydict["model"] = "Mustang GT"
mydict["New_year"] = 2020
mydict["price"] = 35000

print (mydict)
```

****Mutable vs Immutable Objects****  

***Mutable***  (List)
* Same data stores in different location
* Different data stores in different location  

***Immutable***  (Tuple)
* Same data stores in same location
* different daya stores in different location




```python
print ("--------Mutable---------")

a = [10, 20, 30, 40, 50]
b = [10, 20, 30, 40, 50]

print (a, type(a), id(a))
print (b, type(b), id(b))


print ("\n--------Immutable---------") 

a = 100
b = 100
c = 101
print (a, type(a), id(a))
print (b, type(b), id(b))
print ( c, type(c), id(c))

a1 = (10, 20, 30, 40, 50)
b1 = (10, 20, 30, 40, 50)

print (a1, type(a1), id(a1))
print (b1, type(b1), id(b1))
```
