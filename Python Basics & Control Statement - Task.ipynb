***35.	Write a program using while loop to check whether a number is a palindrome.***

```python
num = int(input("Enter a number to check the number is a palindrome or not:"))
num2 = num
reversed = 0
while num > 0:
    digit = num % 10
    reversed = (reversed * 10) + digit
    num //= 10
if num2 == reversed:
    print (f"{num2} is a palindrome")
else:
    print(f"{num2} is not a palindrome")
```
