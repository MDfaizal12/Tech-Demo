***Character - Frequency***

```python
input = "Hello World"
newstr = ""
for each in input:
    if each.isalpha():
        newstr = newstr+each.lower()
print(newstr)
mydict = {}
for x in newstr:
    if x not in mydict.keys():
        mydict[x] = 1
    else:
        mydict[x] += 1
print (mydict)
```

```python
def character(mystr):
    newstr = ''
    mydict = {}

    for each in mystr:
        if each.isalnum():
            newstr = newstr + each.lower()
    for x in newstr:
        if x not in mydict.keys():
            mydict[x] = 1
        else:
            mydict[x] += 1
    return mydict

print (character("9342068032"))
```
