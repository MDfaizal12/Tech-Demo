3.Create a class Rectangle with attributes length and width. Add methods to calculate area and 
perimeter. 

```python
class Rectangle:
    length = 0
    width = 0
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2 * (self.length + self.width)
value1 = Rectangle()
value1.length = 5
value1.width = 3
print("Area:", value1.area())
print("Perimeter:", value1.perimeter())
```

5.Define a class Circle with a class variable pi = 3.14. Add a method to calculate the area. 


```python
class circle:
    pi = 3.14
    def area(self, radius):
        return self.pi * radius * radius
circle1 = circle()
radius = 4
print("Area of circle:", circle1.area(radius))
```

8.Implement a class Library with methods to add_book(title), remove_book(title), and 
display_books(). 

```python
class library:
    books = []
    def add_book(self, title):
        self.books.append(title)
        print(f"Book '{title}' added to the library.")
    def remove_book(self, title):
        if title in self.books:
            self.books.remove(title)
            print(f"\nBook '{title}' removed from the library.")
        else:
            print(f"Book '{title}' not found in the library.")
    def display_book(self):
            if self.books:
                print("\navailable books in the library:")
                for book in self.books:
                    print(f"- {book}")
            else:
                print("No books available in the library.")
lib = library()
lib.add_book("The Great Gatsby")    
lib.add_book("To Kill a Mockingbird")
lib.add_book("1984")
lib.display_book()
lib.remove_book("The Great Gatsby")
    
```

9.Create a class Calculator with methods add, subtract, multiply, and divide. Demonstrate with 
objects. 

```python
class Calculator:
    def add(self, a, b):
        return a + b
    def subtract(self, a, b):
        return a - b
    def multiply(self, a, b):
        return a * b
    def divide(self, a, b):
        if b != 0:
            return a / b
        else:
            return "Error: Division by zero"

calc = Calculator()
print("Addition:", calc.add(10, 5))
print("Subtraction:", calc.subtract(10, 5)) 
print("Multiplication:", calc.multiply(10, 5))
print("Division:", calc.divide(10, 5))
```
