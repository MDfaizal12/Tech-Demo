****Data Types****
* Integer - 123
* Float - 1.0, 1e10
* Complex - 1+j0
* String "Faizal", "123"
* Boolean - True and Flase

```python
#Integer
a = 1009
print("Integer =",a, type(a))

#Float
b = 10.09
print("Float =",b, type(b))

#complex    
c = 10 + 9j
print("Complex =",c, type(c))

#String
d = "Faizal"
print("String =",d, type(d))

#Boolean
e = True
print("Boolean =",e, type(e))
```

****Escape Sequences****
* \ - for newline in editor
* \\ - prints \
* \’– prints ‘
* \”– prints “
* \n – prints a new line
* \t – prints a tab


```python
print("Mohamed \\Faizal\\") 
print("Mohamed\\Faizal")
print("Mohamed \nFaizal")
print("Mohamed \tFaizal")
print("Mohamed \bFaizal") 
print("Mohamed \'Faizal")
print("Mohamed \"Faizal\"")

```
