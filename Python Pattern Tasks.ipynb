**1.Print the below pattern (02/09/2026)**  

&&&&&  
&&&&  
&&&  
&&  
&

```python
row = int(input("enter a number to be row :")) #input = 5
symbol = input("enter a symbol:") #input = &
for i in range(row, 0, -1):
    for j in range(1, i+1 ):
        print(symbol,end = '')
    print()
```

**2.Print the below pattern (2/9/2026)**  
Enter a number between 0-9 : 9  
1  
22  
333  
4444  
55555  
666666  
7777777  
88888888  
999999999

```python
row = int(input("Enter a number between 0-9 :"))

for i in range(1, row+1, 1):
    for j in range(i):
        print(i, end='')
    print()
```

**3.Print the below pattern (10/09/2026)**   
_____/#/  
____/##/    
___/###/  
__/####/  
_/#####/  
/######/  

```python
row = int(input("enter a number to be row :")) #input = 5
symbol = input("enter a symbol:") #input = #
for i in range(1,row+1):
    for j in range(row-i ):
        print(" ",end = '')
    for k in range(i):
        print(symbol, end = '')
    print()
```

**5.Print below pattern (15/09/2026)**

```python
"""
*********
 *******
  *****
   ***
    *
"""
```

```python
row = int(input("Enter a number to be rowed:")) 
symbol = input("Enter a symbol:")

for i in range (row):
    for j in range(i):
        print(" ", end ='')
    for k in range(2*(row-i)-1):
        print(symbol, end = '')
    print()
```
