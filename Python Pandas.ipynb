***Series***   

A 1-dimensional labeled array capable of holding any data type

```python
car_name = pd.Series(["BMW", "Mercedes", "Opel", "Mazda", "Toyota", "Honda", "Ford", "Fiat", "Renault", "Peugeot"])
car_name
```

***DataFrame***  

A 2-dimensional labeled data structure with columns and rows and more then 2 series can form as DataFrame

```python
car_color = pd.Series(["Red", "Blue", "Green", "Black", "White", "Silver", "Yellow", "Orange", "Purple", "Brown"])
car_data = pd.DataFrame({"Car Name": car_name, "Car Color": car_color})
car_data
```

```python
#Add column to the DataFrame
income = pd.Series([50000, 60000, 70000, 80000,])
income
```

```python
new_df["Salary"] = income
new_df
```

```python
#fillna() function is used to fill the missing values in a DataFrame or Series with a specified value or method. It can be used to replace NaN (Not a Number) values with a specific value, forward-fill, backward-fill, or use other methods to fill the missing data.

new_df.fillna(90000, inplace=True)  
new_df
```

**Delete**  

Delete a column using a drop dunction

```python
#TO delete a column from the DataFrame, you can use the drop() function. The drop() function takes the name of the column to be deleted as an argument and removes it from the DataFrame. You can also specify the axis parameter to indicate that you want to delete a column (axis=1) instead of a row (axis=0). Here's an example:
new_df.drop(["Salary"], axis=1, inplace=True)
new_df
```
