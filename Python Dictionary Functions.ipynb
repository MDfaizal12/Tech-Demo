## **Dictionary Function**

```python
mydict = {"Name":"Mohamed Faizal", "city":"Chennai", "State":"Tamil Nadu", "age":22}
```

```python
# keys()
mydict = {"Name":"Mohamed Faizal", "city":"Chennai", "State":"Tamil Nadu", "age":22}

print(mydict)  
print()
print(list(mydict.keys()))
print()
for i in mydict.keys():
    print (i)
```

```python
# values()

mydict = {"Name":"Mohamed Faizal", "city":"Chennai", "State":"Tamil Nadu", "age":22}

print(mydict)
print("------------------")
print(list(mydict.values()))
print('------------------')
for i in mydict.values():
    print(i)
```

```python
# items()
mydict = {"Name":"Mohamed Faizal", "city":"Chennai", "State":"Tamil Nadu", "age":22}

print(list(mydict.items()))
print('------------------')
for i in mydict.items():
    print(f"{i[0]}\t{i[1]}")
```

```python
# List = pop(), pop(index) - if you give empty value inside the () will delete thr last index of the list
# set = pop() - it will delected any random values in the set 
# dictionary = pop(keys), popitem()

mydict = {"Name":"Mohamed Faizal", "city":"Chennai", "State":"Tamil Nadu", "age":22}

mydict.pop("age")
print (mydict)

print('--------------')

mydict.popitem()
print(mydict)
```

```python
#clear()
mydict = {"Name":"Mohamed Faizal", "city":"Chennai", "State":"Tamil Nadu", "age":22}

print(mydict)
print('---------------')
mydict.clear()
print(mydict)
```

```python
# update()

mydict = {"Name":"Mohamed Faizal", "city":"Chennai", "State":"Tamil Nadu", "age":22}

print(mydict)

print('----------------')

newdict = {"Name":"Mishal", "city":"Beach town", "State":"Tamil Nadu", "age":27}

mydict.update(newdict)

print(mydict)
```

```python
# get() 

mydict = {"Name":"Mohamed Faizal", "city":"Chennai", "State":"Tamil Nadu", "age":22}

print(mydict)
print('----------------')
print(mydict.get('age'))
print(mydict.get(500))
```

```python
# copy()

mydict = {"Name":"Mohamed Faizal", "city":"Chennai", "State":"Tamil Nadu", "age":22}
newdict = mydict.copy()
mydict["work"] = "IT"
newdict["work"] = "Gen AI"
print(f"mydict = {mydict}, {type(mydict)}, {id(mydict)}")


print(f"newdict={newdict}, {type(newdict)},{id(newdict)}")
```

### ***Nested Dictionary***

```python
mydict = {"Name":"Mohamed Faizal", "city":["Chennai","kovai","mumbai"], "State":"Tamil Nadu", "age":22, "year":(2000,2001,2002)}

print(mydict, type(mydict))
print("-----------------------------------------")
print(mydict.keys())
print("-----------------------------------------")
print(mydict.get("Name"))
print("-----------------------------------------")
print(mydict.get("city"))
print("-----------------------------------------")
print(mydict.get("city")[2][3:])
print("-----------------------------------------")
print(mydict.get("year"))
print("-----------------------------------------")
print(mydict.get("year")[2])
```
