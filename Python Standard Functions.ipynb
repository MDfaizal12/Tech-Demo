**Python Standard Function**

```python
#len()

x = "hellowroldhelloworld"
print(len(x))
```

```python
#max()
#min()
#sum()

x = [3,2,4,5,6,7,8,-9,-10,100,244,-34]

print(x)
print(min(x))
print(max(x))
print(sum(x))
```

```python
#abs()

a = 24.44
a1 = abs(a)
print(a1)
```

```python
#pow()
 
print(10**3)
print(pow(10,3))
```

```python
#round()

x = 1009.10
y = 1009.51
print(round(x))
print(round(y))
```

```python
#slice()
x = [1,2,4,5,666,7,8]
print(x[1:5])
print(x[slice(1,5)])
```

```python
#sorted()

x = [-4,-1,-2,-5,-6,-7,-3,0,1,2,3,5,7,4,6]

print(x)
print(sorted(x))
print(sorted(x, reverse = True))
print()
```

```python
#chr()

for i in range(33,127):
    print(chr(i), end =" ")
```

```python
#enumerate()

x = [2,3,5,6,7,8,9,5,5,5,8,9,0,4,3,2,6,3,0,2,1,1,5]
print(x)  

mylist = list(enumerate(x))

for i in mylist:
    print(i)
```

```python
#enumerate()

x = [2,3,5,6,7,8,9,5,5,5,8,9,0,4,3,2,6,3,0,2,1,1,5]

search = 2

mylist = list(enumerate(x))

for i in mylist:
    if i[1] == search:
        print(f"the element found in:({i[0]} {i[1]})")
```

```python
x = "Hello world" 
search = 'l'
mylist = list(enumerate(x))
for i in mylist:
    if i[1] == search:
        print(f"The element  found at = ({i[0]} {i[1]})")
```
