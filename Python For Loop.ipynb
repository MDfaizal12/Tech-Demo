#### *****for loop*****  
For loop is used for iterating over a sequence (that is either a list, a  
tuple, a dictionary, a set, or a string). With the for loop we can execute  a set 
of statements, once for each item in a list, tuple, set etc.  

***Syntax***  
for \<variable_name> in \<str/list/tuple/set/dict>:

```python
mystr = "Hello World"
for x in mystr:
    print(x, end='')
```

```python
mylist = ["apple", "banana", "cherry"]
for x in mylist:
    print(x, end=',')
```

```python
mytuple = ("apple", "banana", "cherry")
for x in mytuple:
    print(x, end=',')
```

****range()****  
range is a python standard function used for numbers 

```python
for x in range(6):
    print(x, end=',')
```

```python
for x in range(2, 10+1):
    print(x, end=',')
```

```python
num = int(input("Enter a number: "))
for x in range(0,num+1):
    print(x, end=',')
```

```python
# Jumping numbers in a for loop
for x in range(0,10+1,3):
    print(x, end=',')
```

```python
# reverse order in a for loop
for x in range(10, 0, -1):
    print(x, end=',')
```

***for with else***

```python
for x in range(6):
    print(x, end=',')
else:
    print("\nFinally finished!")
```

```python
for x in "HelloWorld":
    if x == " ":
        break
    print(x, end='')
else:
    print("\nFinally finished!")
```

```python
# find if the number is divisible by 5
mylist = [7, 8, 10, 12, 15, 20, 25, 30]
for each in mylist:
    if each % 5 == 0:
        print(each, "is divisible by 5")
        break
else:
    print(each, "is not divisible by 5")
```

***Nested Loop***

```python
for x in range(5):
    for y in range(1,5+1):
        print( y, end='')
    print()
```

```python
color = ["red", "green", "blue"]
animal = ["cat", "dog", "rabbit"]
for x in color:
    print("====================")
    for y in animal:
        print(x, y)
    print("====================")
```
