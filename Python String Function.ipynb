### **Python String Function**  
String have positive index and negative index for a sting

```python
a = "Mohamed Faizal"
print(a)
print(a[2])
print(a[-1])
print(a[0:9])
print(a[:9])
print(a[0:15:3])
print(a[::3])
print(a[::-1])
```
