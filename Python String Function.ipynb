**Interview coding Task**  
1. Find weather the string is palindrome 
1. Input: racecar  
   Output: True  
2. Input: MalaYa lAm  
   Output: True  
3. Input: Sir, I demand, I am a maid named Iris  
   Output: True
4. Input: Red roses run no risk. Sir, on Nurse's Order.  
   Output: True

```python
mystr = "racecar"
rev = mystr[::-1]

print(mystr)
print(rev)

if mystr == rev:
    print(True)
else:
    print(False)
```

```python
mystr = "Red roses run no risk. Sir, on Nurse's Order."
newstr = ''
for each in mystr:
    if each.isalpha():
        newstr += each.lower()
print(newstr)
if newstr == newstr[::-1]:
    print(True)
else:
    print(False)
```

```python
# def function

def palin(mystr):
    newstr =''
    for each in mystr:
        if each.isalpha():
            newstr += each.lower()

    if newstr == newstr[::-1]:
        print(True)
    else:
        print(False)

print(palin("Red roses run no risk. Sir, on Nurse's Order."))
print(palin("Sir, I demand, I am a maid named Iris"))
print(palin("MalaYa lAm"))
```
