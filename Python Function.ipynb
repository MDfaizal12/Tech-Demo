***def \<function_name>():***

def 

```python
def demo_fun():
    print("I\'m happy")
def demo_fun1():
    print("I\'m sad")
demo_fun()
print()
demo_fun1()
print()
```

### **Python Function with Parameters**  

```python
def demo_fun2(x,y): #x,y ==> Parameters
    print("x:", x)
    print("y:", y)
    print("the sum of:", x+y)
demo_fun2(20,40)
print()
```

### **Python Function with defalt Parameters**  

```python
def demo_fun3(x=0, y=0):
    print("x:",x)
    print("y:",y)
    print("the sum is :", x+y)
demo_fun3()
print()
demo_fun3(30,70)
print()
```

### **Return**

```python
def demo_fun4(x=0, y=0):
    print("x =", x)
    print("y =", y)
    print("The sum of:",x+y)
ans = demo_fun4(100,200)
print(ans)
```

```python
def demo_fun4(x=0, y=0):
    return x+y
def demo_fun5(x=0, y=0):
    return x*y
ans = demo_fun5(demo_fun4(70, 30), demo_fun5(30, 50))
print(ans)
```
