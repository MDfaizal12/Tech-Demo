### **Scope**  

*types of scope*  
* local scope - Local scope can created multiple times and change of values 
* Global scope - Global scope can create only once and can't change the values 

```python
x = 10 #global scope
y = 200

print(f"x in  Global scope = {x}")
print(f"Y in Global scope = {y}")
print()
def myscope(x=0, y=0): #local scope
    print(f"x in  Global scope = {x}")
    print(f"Y in Global scope = {y}")
myscope()
print()
myscope(100,1009)
```

### **recursion**  
A function is called within the same function is called Recursion. 
Recursion is  a common mathematical and programming concept. It means 
that a  function calls itself. 

```python
def myfun(x,y):
    return x * y
def mysad(x, y):
    return x // y

print (mysad(myfun(10,4),myfun(2,4)))
```

```python
def fact(num):
    nums = 1
    for i in range(1, num+1):
        nums = nums*i
    return nums

print(fact(5))
```

```python
# same factorial in recursion function

def fact(num):
    if num == 1:
        return 1
    else:
        return num * fact(num-1) #used to call the fucntion ifself

print(fact(5))
```
