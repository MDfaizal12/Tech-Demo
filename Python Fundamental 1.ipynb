***Type casting***

```python
a = 1009
a1 = float(a)
a2 = str(a)
a3 = bool(a)
a4 = complex(a)
print ("The value of a is: ", a)
print ("The value of a1 is: ", a1)
print ("The value of a2 is: ", a2)  
print ("The value of a3 is: ", a3)
print ("The value of a4 is: ", a4)
```

```python
b = "Faizal"
b2 = bool(b)
print ("The value of b is: ", b, type(b))
print ("The value of b2 is: ", b2, type(b2))

```
