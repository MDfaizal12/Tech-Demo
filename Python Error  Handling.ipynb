### ***raise - keyword***  
The raise keyword is used to raise an exception on own by the user

```python
print("Few lines above")

try:
    num = int(input("Enter a  number:"))

    if num <=10:
        raise ValueError
    elif num >= 50:
        raise ZeroDivisionError
    else:
        print("Just for fun, carryon")
except ValueError:
    print("Hi the give interger is correct but i raised a error")
except ZeroDivisionError:
    print("Same I raised a error ")
```

### ***Task 1: List Index and input Error***  
* Get a list of numbers from the user
* Asks the user to enter an index
* Use nested try-except blocks to handle
* ValueError and IndecError

```python
try:
    numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

    try:
        index = int(input("Enter an index: "))
        print(f"Number at index {index}: {numbers[index]}")

    except ValueError:
        print("Please enter a valid integer index.")
    except IndexError:
        print("Index is out of range.")

except ValueError:
    print("Please enter only valid numbers.")
```

### ***Task 2: Dictionary Access***  
Create a student dictionary  
student = {
    "name":"Manju"
    "age":25
    "course":"Python"
}  
Ask the user for a key and display its value.  
use nested exception handling to handle: Invalid Input and keyError  

```python
student = {
    "name": "Manju",
    "age": 25,
    "course": "Python"
}
while True:
    try:
        try:
            key = input("Enter a student key: ")

            if not key:
                raise ValueError("Input cannot be empty.")

            print(f"{key}: {student[key]}")
            break

        except ValueError as e:
            print(e)
        except KeyError:
            print("Key not found. Please enter name, age, or course.")

    except Exception as e:
        print(f"An unexpected error occurred: {e}")
        break
```
