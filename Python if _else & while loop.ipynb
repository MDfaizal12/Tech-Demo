### ***if else***  
* if
* elif
* else  
if else is a decision statement where choose of output  

***rules***
* if (only once per code block)
* elif (as many time as per code block)
* else ( Only once per code block)  

**syntax**
* if \<condition>:
* elif \<condition>:
* else:


***if statement***

```python
a = 1009
if a > 1000:
    print("a is greater than 1000")
print("This line will always be printed")
```

***if else Statement***

```python
a = 1000
if a > 1000:
    print("a is greater than 1000")
else:
    print("a is not greater than 1000")
print("This line will always be printed")
```

***if elif else statement***

```python
a = 1000
if a > 1000:
    print("a is greater than 1000") 
elif a == 1000:
    print("a is equal to 1000")
else:
    print("a is less than 1000")    
print("This line will always be printed")
```

```python
a = 1000
if  a > 1000:
    print("a is greater than 1000")
    if a > 1006:
        print("a is greater than 1006")
    else:
        print("a is less than 1006")
elif a >= 1000:
    print("a is equal to 1000")
    if a < 1006:
        print("a is less than 1006")
    else:
        print("a is greater than 1006")
else:
    print("a is less than 1000")
print("Code execution completed")
```

***Funtime Game***

```python
score = 100
secret_nnumber = 50
guess = int(input("Enter your guess: "))
if guess == secret_nnumber:
    print("correct")
    score += 10
    print("your score is", score)
elif guess > secret_nnumber:
    print("too high")
    score -= 10
    print("your score is", score)
else:
    print("too low")
    score -= 10
    print("your score is", score)
```

### ***While Loop***

while loop is a condition  statement, is a control flow statement used to repeatedly execute a block of code as long as a specified condition remains true

**Syntax**
* while \<condition>:

```python
a = 100
while a >= 0:
    print(a, end = " ")
    a -= 10
print("\nloop over")
```

```python
a = 1
while a <= 20:
    if a % 3 == 0:
        print(a, end = " ")
        a += 1
        continue
    a += 1
print("\nloop over")
```

***Nested While Loop***

```python
a = 1
while a <= 5:
    b = 1
    while b <= 5:
        print(b, end = "")
        b += 1
    a += 1
    print()
```

***Task of Day 3***  
Using a nested while loop condition while fetching input from the users, and printing the output as rows and columns. 

```python
row = int(input("enter number of rows: "))
column = int(input("enter number of columns: "))

a = 1
while a <= row:
    b = 1
    while b <= column:
        print(b, end = "")
        b += 1
    a += 1
    print()
```
