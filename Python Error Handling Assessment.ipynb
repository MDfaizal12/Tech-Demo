***4. What  is the difference between Exception and BaseException in Python?***  

Exception:
* Exception is a subclass of BaseException, its main purpose are runtime bugs, invalid inputs, and recoverable failures  
* Exception have ValueError, KeyError, TypeError, etc.

BaseException:  
* BaseException is the high level root of all exceptions and it's main purpose is for signals intended to exit or shut down python cleanly.
* BaseException have SystemExit, KeyboardInterrupt, generatorExit, Exception.
