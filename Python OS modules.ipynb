```python
import os
def findallfile (filename, path):
    match = []
    for root, folder, files in os.walk(path):
        for file in files:
            if file == filename or file.endswith(filename):
                full_path = os.path.join(root,file)
                match.append(full_path)
    return match,"\n"

print(findallfile(".pdf","R:\\certificates"))
```

```python
import os
def find(name, path):
    match = []
    for root, folder, files, in os.walk(path):
        if name in files:
            match.append(os.path.join(root,name))
    return match

for path in find("Faizal.pdf","R:\\certificates"):
    print(path)
print()
```
